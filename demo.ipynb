# KODEX demo — predict in milliseconds, with calibrated uncertainty and a trust gate
The Kronos Family of Codes. One contract: `predict(x) -> (y, uncertainty, in_domain)`.
Nothing here is published; representative-fidelity CGYRO (μ=400).

```python
import kronos_ml as K
print(K.__version__, '·', len(K.SURROGATES), 'codes')
K.fleet(phase=1)
```

## KYRO — CGYRO turbulence transport in ~0.3 ms

```python
p = K.run('KYRO', [3.5, 1.6])   # a/L_T, shear
print(f'Q_i = {p.y:.0f} ± {p.uncertainty:.0f}  (in_domain={p.in_domain})')
p.as_tagged('CGYRO (nonlinear gyrokinetic)')   # provenance travels with the value
```

## KGATE — the model-free trust boundary

```python
g = K.get('KGATE')
safe, ok = g.clamp([2.0, 0.3, 9.0, 5.0, 0.9, 60.0])   # over-limit command
print('projected onto envelope:', safe.round(2), '· in_domain=', g.in_domain([2.0,0.3,9.0,5.0,0.9,60.0]))
```
